# ЖОБА ПАСПОРТЫ

**Жоба тақырыбы:** Titanic деректері негізінде жолаушылардың аман қалуын болжау.

**Мәселе:** Titanic кемесіндегі жолаушылардың барлығы бірдей апаттан аман қалмады. Олардың аман қалуына жасы, жынысы, билет класы және басқа факторлар әсер етті.

**Мақсаты:** Машиналық оқыту әдістерін пайдаланып, жолаушылардың аман қалу ықтималдығын болжау.

**Деректер көзі:** Kaggle – Titanic Dataset.

**Модель түрі:** Machine Learning, Random Forest Classifier және Dummy Classifier.

**Сапа метрикасы:** F1-score (macro).

**Қабылдау шегі:** F1-score ≥ 0.75.

**Қолданылатын технологиялар:** Python, Pandas, Scikit-learn, Google Colab.

**Күтілетін қиындықтар:** Бос мәндерді өңдеу, деректерді тазалау, модель сапасын арттыру және нәтижелерді салыстыру.

**Күтілетін нәтиже:** Жолаушылардың аман қалуын болжайтын машиналық оқыту моделі.

```python
import pandas as pd
import seaborn as sns

df = sns.load_dataset("titanic")

print("Деректер өлшемі:")
print(df.shape)

print("\nАлғашқы 5 жол:")
print(df.head())

print(df.info())

print(df.isna().sum())

print(df.describe())
```

```python
from sklearn.model_selection import train_test_split

# Қажетті бағандарды таңдау
df = df[["survived", "pclass", "sex", "age", "fare", "sibsp", "parch"]]

# Бос мәндерді өңдеу
df = df.dropna()

# Мәтіндік мәндерді санға айналдыру
df["sex"] = df["sex"].map({"male": 0, "female": 1})

# Мақсатты баған
target = "survived"

# Train және қалған деректер
train, temp = train_test_split(
    df,
    test_size=0.3,
    random_state=42,
    stratify=df[target]
)

# Validation және Test
val, test = train_test_split(
    temp,
    test_size=0.5,
    random_state=42,
    stratify=temp[target]
)

print("Train:", len(train))
print("Validation:", len(val))
print("Test:", len(test))
```

```python
from sklearn.dummy import DummyClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score, accuracy_score

# Деректерді бөлу
target = "survived"

Xtr = train.drop(columns=target)
ytr = train[target]

Xv = val.drop(columns=target)
yv = val[target]

# Baseline модель
base = DummyClassifier(strategy="most_frequent")
base.fit(Xtr, ytr)

# Random Forest моделі
rf = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)

rf.fit(Xtr, ytr)

# Болжау
pred_base = base.predict(Xv)
pred_rf = rf.predict(Xv)

# Нәтижелер
print("Baseline F1:", f1_score(yv, pred_base, average="macro"))
print("Random Forest F1:", f1_score(yv, pred_rf, average="macro"))

print("\nBaseline Accuracy:", accuracy_score(yv, pred_base))
print("Random Forest Accuracy:", accuracy_score(yv, pred_rf))
```
